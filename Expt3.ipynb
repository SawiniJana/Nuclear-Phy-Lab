### Part A

```python
import numpy as np
import matplotlib.pyplot as plt

# Data
theory = np.array([-4.54, -1.00, -2.20, -10.00])
measured = np.array([-4.59, -1.03, -2.23, -10.02])

# Linear fit
coeffs, cov = np.polyfit(theory, measured, 1, cov=True)
m, c = coeffs

# Uncertainty (standard deviation)
m_err = np.sqrt(cov[0,0])
c_err = np.sqrt(cov[1,1])

# R^2 calculation
y_fit = m*theory + c
ss_res = np.sum((measured - y_fit)**2)
ss_tot = np.sum((measured - np.mean(measured))**2)
r2 = 1 - (ss_res / ss_tot)

# Smooth line
x = np.linspace(min(theory)-1, max(theory)+1, 100)

# Plot
plt.figure()

plt.scatter(theory, measured, label="Data Points")

plt.plot(x, m*x + c, linestyle='--', color="black",
         label=f"Linear Fit (R² = {r2:.3f})")

# Labels
plt.xlabel(r"Theoretical Gain")
plt.ylabel(r"Measured Gain")
plt.title("Measured vs Theoretical Gain")

# Annotation (inside plot)
plt.text(min(theory), max(measured),
         f"Slope = {m:.3f} ± {m_err:.3f}\nIntercept = {c:.3f} ± {c_err:.3f}",
         fontsize=10, verticalalignment='top')

# Grid and legend
plt.grid()
plt.legend()

plt.show()
```
